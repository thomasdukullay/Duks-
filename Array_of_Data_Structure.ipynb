```python
my_array = [10, 20, 30, 40, 50]
total = 0

for num in my_array:
    total += num

print("The total sum of elements is:", total)
```

```python
my_array = []
print("Enter numbers to add to the array. Press Enter without typing a number to finish.")

while True:
    user_input = input("Enter a number: ")
    if user_input == "":
        break
    my_array += [int(user_input)] # Manually add to array

total = 0
for num in my_array:
    total += num

print("The array is:", my_array)
print("The sum of the entered elements is:", total)
```

```python
my_array = [12, 45, 7, 89, 34, 2]

if len(my_array) > 0:
    largest = my_array[0]
    for num in my_array:
        if num > largest:
            largest = num
    print("The largest number is:", largest)
else:
    print("The array is empty.")
```

```python
my_array = [12, 45, 7, 89, 34, 2]

if len(my_array) > 0:
    smallest = my_array[0]
    for num in my_array:
        if num < smallest:
            smallest = num
    print("The smallest number is:", smallest)
else:
    print("The array is empty.")
```

```python
my_array = [15, 22, 39, 41, 56]
target = 41
found_index = -1

for i in range(len(my_array)):
    if my_array[i] == target:
        found_index = i
        break

if found_index != -1:
    print(f"Number {target} found at index {found_index}.")
else:
    print(f"Number {target} not found in the array.")
```

```python
my_array = [10, 20, 30, 40]
element_to_add = 99
target_index = 2
new_array = []

for i in range(len(my_array) + 1):
    if i == target_index:
        new_array += [element_to_add]
    elif i < len(my_array):
        new_array += [my_array[i]]

my_array = new_array
print("Array after insertion:", my_array)
```
