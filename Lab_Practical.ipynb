```python
import matplotlib.pyplot as plt
import numpy as np

# Secondary School Registration data using dictionaries
JSS1 = {'Male': 45, 'Female': 55}
JSS2 = {'Male': 50, 'Female': 60}
JSS3 = {'Male': 55, 'Female': 65}
SSS1 = {'Male': 60, 'Female': 70}
SSS2 = {'Male': 58, 'Female': 68}

# Combine data for easier plotting
classes = ['JSS1', 'JSS2', 'JSS3', 'SSS1', 'SSS2']
male_counts = [JSS1['Male'], JSS2['Male'], JSS3['Male'], SSS1['Male'], SSS2['Male']]
female_counts = [JSS1['Female'], JSS2['Female'], JSS3['Female'], SSS1['Female'], SSS2['Female']]

x = np.arange(len(classes))  # the label locations
width = 0.5  # Adjusted width for stacked bars

plt.figure(figsize=(12, 6))

# Plotting Male registration (bottom of the stack)
rects1 = plt.bar(x, male_counts, width, label='Male', color='skyblue')
# Plotting Female registration (on top of Male)
rects2 = plt.bar(x, female_counts, width, bottom=male_counts, label='Female', color='lightcoral')

# Add labels, title and custom x-axis tick labels
plt.xlabel('Classes')
plt.ylabel('Number of Students')
plt.title('Secondary School Registration by Class and Gender (Stacked)')
plt.xticks(x, classes)
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Function to add labels on top of bars
def autolabel(rects, base_counts=None):
    for i, rect in enumerate(rects):
        height = rect.get_height()
        if base_counts is not None:
            # For the top segment of a stacked bar, position at total height
            total_height = base_counts[i] + height
            plt.annotate('{}'.format(height), # Display the segment's height
                        xy=(rect.get_x() + rect.get_width() / 2, total_height),
                        xytext=(0, 3),  # 3 points vertical offset
                        textcoords="offset points",
                        ha='center', va='bottom')
        else:
            # For the bottom segment of a stacked bar
            plt.annotate('{}'.format(height),
                        xy=(rect.get_x() + rect.get_width() / 2, height),
                        xytext=(0, 3),  # 3 points vertical offset
                        textcoords="offset points",
                        ha='center', va='bottom')

autolabel(rects1) # Labels for male counts
autolabel(rects2, male_counts) # Labels for female counts, positioned above male counts

plt.tight_layout()
plt.show()
```
